```python
import pandas as pd
import matplotlib.pyplot as plt
import geopandas as gpd
import contextily as cx
import folium
from folium.plugins import HeatMap
```

# Load dataset

```python
dataset_path = "./tsunamis-2026-05-25_16-23-01_+0900.csv"
df = pd.read_csv(dataset_path)
print(df.shape)
df.tail()
```

# Dataset details
01/01/2000 to 15/05/2026

```python
#tsunami dataset details
print("Dataset shape:", df.shape)
print("Columns:", df.columns)
print("Missing values:\n", df.isnull().sum())
```

# Tsunami location map

```python
# tsunami locations using folium
#center the map on the average location of the tsunamis

avg_lat = df['Latitude'].mean()
avg_lon = df['Longitude'].mean()
m = folium.Map(location=[avg_lat, avg_lon], zoom_start=5)
for _, row in df.iterrows():
    if pd.notnull(row['Latitude']) and pd.notnull(row['Longitude']):
        folium.CircleMarker(
            location=[row['Latitude'], row['Longitude']],
            radius=5,
            color='red',
            fill=True,
            fill_color='red'
        ).add_to(m)


m
```

# One tsunami detail

```python
#show details on the ONE biggest tsunamis using "Maximum Water Height (m)"

top_tsunamis = df.sort_values(by='Maximum Water Height (m)', ascending=False).head(1)
print(top_tsunamis)

```
